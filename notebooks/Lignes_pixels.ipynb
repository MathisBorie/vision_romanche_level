# Détection du niveau d'eau

Exécutez les cellules dans l'ordre (ou choisissez **Run All**) pour lancer `Lignes_pixel2.py`, générer le rapport PDF, puis l'afficher ci-dessous.

```python
from pathlib import Path
import subprocess
import sys

# Trouve la racine du dépôt même si le notebook est lancé depuis notebooks/.
repo_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents) if (path / "Lignes_pixels.py").is_file()),
    None,
)
if repo_root is None:
    raise FileNotFoundError("Impossible de trouver Lignes_pixels.py depuis le dossier courant.")

script_path = repo_root / "Lignes_pixels.py"
report_path = repo_root / "Lignes_pixels_report.pdf"

subprocess.run([sys.executable, str(script_path)], cwd=repo_root, check=True)

if not report_path.is_file():
    raise FileNotFoundError(f"Le rapport PDF n'a pas été créé : {report_path}")

print(f"Rapport prêt : {report_path}")
```

## Rapport PDF

Le lecteur ci-dessous permet de parcourir les pages. Le lien ouvre aussi le PDF séparément si le lecteur intégré n'est pas pris en charge.

```python
from IPython.display import IFrame, Markdown, display

pdf_url = f"/files/{report_path.relative_to(repo_root).as_posix()}"
display(Markdown(f"[Ouvrir le PDF dans un nouvel onglet](../{report_path.name})"))
display(IFrame(src=pdf_url, width="100%", height=800))
```
